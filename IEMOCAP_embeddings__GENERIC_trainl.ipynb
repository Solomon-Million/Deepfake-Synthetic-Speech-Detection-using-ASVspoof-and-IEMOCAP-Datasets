### Description
This script extracts generic audio embeddings from the "train" subset of the IEMOCAP dataset using the pretrained Wav2Vec2 model. The embeddings are generated through mean pooling of frame-level representations from the model's last hidden layer, providing a compact and meaningful representation for each audio file. The script associates each embedding with its corresponding emotion label and saves the results in a structured CSV file.

### Workflow

1. **Dataset Extraction and Setup**:  
   - Unzips the IEMOCAP dataset and sets up the directory structure for processing.  
   - Reads the metadata CSV file to map audio file paths to emotion labels.

2. **Model and Processor Initialization**:  
   - Loads the pretrained Wav2Vec2 processor and model from Hugging Face for feature extraction.

3. **Audio Processing and Embedding Extraction**:  
   - Loads each audio file using `librosa` at a sample rate of 16 kHz.  
   - Processes the audio data with the Wav2Vec2 processor and extracts embeddings using the model in inference mode.  
   - Performs mean pooling across frame-level embeddings to create a single fixed-size embedding for each audio file.

4. **Metadata Association**:  
   - Associates each embedding with its corresponding file path, folder, emotion label, and embedding shape.

5. **Save Results**:  
   - Converts the list of embeddings and metadata into a Pandas DataFrame.  
   - Saves the embeddings and metadata to a CSV file for easy access and reuse in downstream tasks.

### Key Features

- **Pretrained Wav2Vec2 Embeddings**: Leverages the robust Wav2Vec2 model to generate meaningful audio embeddings.  
- **Emotion Label Association**: Links each embedding to its respective emotion label from the metadata.  
- **Scalable and Efficient**: Processes large audio datasets efficiently with batch-wise operations and inference-only computations.  
- **Reproducible Output**: Saves the processed embeddings and metadata in a structured CSV format for easy reuse.

This script is ideal for researchers working on emotion recognition or other audio-based classification tasks, providing a structured approach to extract and save embeddings for model training.


```python
# Define path for the new zip file and extraction
zip_path = '/content/drive/MyDrive/Copy of IEMOCAP_new.zip'
extract_path = '/content/iemocap_data'

# Extract Copy of IEMOCAP_new.zip if not already extracted
if not os.path.exists(extract_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)

# Define paths to the CSV and the "train" folder inside the extracted "IEMOCAP_Copied" folder
csv_path = os.path.join(extract_path, 'Copy of IEMOCAP', 'IEMOCAP_Copied', 'iemocap_full_dataset.csv')
audio_folder = os.path.join(extract_path, 'Copy of IEMOCAP', 'IEMOCAP_Copied', 'train')

# Load the CSV file into a dictionary for easy lookup
data = pd.read_csv(csv_path)
csv_files = {row['path'].replace('/', '_').replace('.wav', ''): row['emotion'] for _, row in data.iterrows()}

# Load the Wav2Vec 2.0 large model and processor from Hugging Face
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-large-960h")
model = Wav2Vec2Model.from_pretrained("facebook/wav2vec2-large-960h")

# Function to load audio, process it, and extract embeddings
def extract_embeddings(file_path):
    try:
        # Attempt to load the audio file with librosa at a sample rate of 16 kHz
        audio_input, _ = librosa.load(file_path, sr=16000)
    except Exception as e:
        # If loading fails, print a warning and skip the file
        print(f"Warning: Could not load {file_path}. Skipping. Error: {e}")
        return None, None  # Return None to indicate a problem with this file

    # Prepare the input values for the Wav2Vec 2.0 model
    input_values = processor(audio_input, sampling_rate=16000, return_tensors="pt").input_values

    # Extract embeddings using the model (without gradients as we're just using it for inference)
    with torch.no_grad():
        frame_embeddings = model(input_values).last_hidden_state

    # Mean pooling across frames to get a single embedding vector per audio file
    pooled_embedding = frame_embeddings.mean(dim=1).squeeze()

    # Return the embedding and its shape
    return pooled_embedding.numpy(), pooled_embedding.shape

# Initialize list to store embeddings
embeddings_list = []

# Process each file in the "train" folder
for filename in tqdm(os.listdir(audio_folder)):
    base_filename = filename.replace('.wav', '')

    # Check if the file is listed in the CSV
    if base_filename not in csv_files:
        continue  # Skip if not in CSV

    file_path = os.path.join(audio_folder, filename)

    # Extract embeddings from the WAV file
    emotion = csv_files[base_filename]
    embeddings, embedding_shape = extract_embeddings(file_path)

    # Skip adding to the list if embeddings were not extracted (i.e., file was corrupted)
    if embeddings is None:
        continue

    # Append embeddings, file path, label, and shape info to list
    embeddings_list.append({
        "file_path": base_filename,
        "folder": "train",
        "emotion_label": emotion,
        "embedding_shape": embedding_shape,
        "embedding": embeddings
    })

# Convert list to DataFrame
embeddings_df = pd.DataFrame(embeddings_list)

# Convert embeddings to lists for saving in a CSV
embeddings_df['embedding'] = embeddings_df['embedding'].apply(lambda x: x.tolist())

# Save the DataFrame to CSV
embeddings_csv_path = "/content/iemocap_embeddings_train.csv"
embeddings_df.to_csv(embeddings_csv_path, index=False)

# Display the DataFrame in Colab
print(embeddings_df)

print(f"Embeddings and metadata saved to {embeddings_csv_path}")
```
