###Overview:
This script processes raw IEMOCAP audio embeddings stored in CSV files by converting string-formatted embedding data into structured numerical arrays. The cleaned embeddings are saved as .npy files for efficient storage and rapid loading during machine learning model training and evaluation. The workflow ensures that embeddings for training, validation, and test datasets are standardized, accessible, and ready for downstream tasks without requiring additional preprocessing

```python
import pandas as pd
import numpy as np
import os

# Function to clean and convert the 'embedding' column from string representation to a list of floats
def process_embedding_column(df, column_name='embedding'):
    df[column_name] = df[column_name].apply(
        lambda x: [float(i.strip()) for i in x.strip('[]').split(',') if i.strip()]
    )
    return np.array(df[column_name].tolist())

# Paths to your Embedding data CSV files
embedding_train_csv_path = '/content/drive/MyDrive/IEMOCAP_embeddings_NEW_bce/iemocap_embeddings_train_NEW_bce.csv'
embedding_val_csv_path = '/content/drive/MyDrive/IEMOCAP_embeddings_NEW_bce/iemocap_embeddings_validate_NEW_bce.csv'
embedding_test_csv_path = '/content/drive/MyDrive/IEMOCAP_embeddings_NEW_bce/iemocap_embeddings_test_NEW_bce.csv'

# Folder to save the processed Embedding data
save_folder = '/content/drive/MyDrive/IEMOCAP_embeddings_NEW_bce'
os.makedirs(save_folder, exist_ok=True)  # Ensure the directory exists

# Paths to save the processed Embedding data
processed_train_embedding_path = os.path.join(save_folder, 'processed_embedding_train.npy')
processed_val_embedding_path = os.path.join(save_folder, 'processed_embedding_validate.npy')
processed_test_embedding_path = os.path.join(save_folder, 'processed_embedding_test.npy')

# Process and save Embedding Train Data
df_train_embedding = pd.read_csv(embedding_train_csv_path)
processed_train_embedding = process_embedding_column(df_train_embedding)
np.save(processed_train_embedding_path, processed_train_embedding)
print("Processed Embedding train data saved successfully in Google Drive!")

# Process and save Embedding Validation Data
df_val_embedding = pd.read_csv(embedding_val_csv_path)
processed_val_embedding = process_embedding_column(df_val_embedding)
np.save(processed_val_embedding_path, processed_val_embedding)
print("Processed Embedding validation data saved successfully in Google Drive!")

# Process and save Embedding Test Data
df_test_embedding = pd.read_csv(embedding_test_csv_path)
processed_test_embedding = process_embedding_column(df_test_embedding)
np.save(processed_test_embedding_path, processed_test_embedding)
print("Processed Embedding test data saved successfully in Google Drive!")
```

###Overview:
This script loads processed IEMOCAP embeddings from .npy files and converts them into PyTorch tensors for seamless integration into deep learning workflows. The tensors are saved in .pt format, which is optimized for PyTorch operations and efficient loading. The script ensures embeddings for training, validation, and test datasets are properly converted and stored, with checks to confirm their shapes and successful saving.

```python
import numpy as np
import torch
import os

# Define the folder path for embeddings
embedding_folder_path = '/content/drive/MyDrive/IEMOCAP_embeddings_NEW_bce'

# Load the processed embedding data for train, validate, and test datasets from the folder
embedding_train = np.load(os.path.join(embedding_folder_path, 'processed_embedding_train.npy'))
embedding_validate = np.load(os.path.join(embedding_folder_path, 'processed_embedding_validate.npy'))
embedding_test = np.load(os.path.join(embedding_folder_path, 'processed_embedding_test.npy'))

# Convert the embedding data into PyTorch tensors
embedding_train_tensor = torch.tensor(embedding_train, dtype=torch.float32)
embedding_validate_tensor = torch.tensor(embedding_validate, dtype=torch.float32)
embedding_test_tensor = torch.tensor(embedding_test, dtype=torch.float32)

# Define save paths for the processed tensors
train_file_path = os.path.join(embedding_folder_path, 'embedding_train_tensor.pt')
validate_file_path = os.path.join(embedding_folder_path, 'embedding_validate_tensor.pt')
test_file_path = os.path.join(embedding_folder_path, 'embedding_test_tensor.pt')

# Save the embedding tensors to the defined paths
torch.save(embedding_train_tensor, train_file_path)
torch.save(embedding_validate_tensor, validate_file_path)
torch.save(embedding_test_tensor, test_file_path)

# Print confirmation and check tensor shapes
print(f"Embedding train tensor saved at: {train_file_path}")
print(f"Embedding validate tensor saved at: {validate_file_path}")
print(f"Embedding test tensor saved at: {test_file_path}")

# Check tensor shapes for confirmation
print("Embedding train tensor shape:", embedding_train_tensor.shape)
print("Embedding validate tensor shape:", embedding_validate_tensor.shape)
print("Embedding test tensor shape:", embedding_test_tensor.shape)
```
