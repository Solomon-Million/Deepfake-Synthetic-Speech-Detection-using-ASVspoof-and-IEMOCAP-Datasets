```python
import os
import random
import shutil

# Base directory for the original IEMOCAP dataset
iemocap_base_dir = '/content/drive/My Drive/MENG/Deep Learning/Project/IEMOCAP_full_release'

# Collect all the .wav files from all 5 sessions
iemocap_files = []
for session in range(1, 6):  # Session1 to Session5
    session_wav_dir = os.path.join(iemocap_base_dir, f'Session{session}', 'sentences', 'wav')

    # Walk through all subfolders (impro, script, etc.) and collect .wav files
    for root, dirs, files in os.walk(session_wav_dir):
        for file in files:
            if file.endswith('.wav'):
                file_path = os.path.join(root, file)
                iemocap_files.append(file_path)

# Set the target sample size
total_sample_size = 10039
train_sample_size = int(0.7 * total_sample_size)  # 70% for training
dev_sample_size = int(0.15 * total_sample_size)   # 15% for validation
eval_sample_size = total_sample_size - train_sample_size - dev_sample_size  # Remaining 15% for testing

# Randomly sample files from the dataset
sampled_iemocap_files = random.sample(iemocap_files, total_sample_size)

# Split into train, validation, and test sets
train_files = sampled_iemocap_files[:train_sample_size]
validate_files = sampled_iemocap_files[train_sample_size:train_sample_size + dev_sample_size]
test_files = sampled_iemocap_files[train_sample_size + dev_sample_size:]

# Directory to save the new dataset in Google Drive
new_iemocap_dir = '/content/drive/My Drive/MENG/Deep Learning/Project/IEMOCAP_NewSplit'  # Adjust path as needed

# Create directories for train, validation, and test sets in the new dataset
train_dir = os.path.join(new_iemocap_dir, 'train')
validate_dir = os.path.join(new_iemocap_dir, 'validate')
test_dir = os.path.join(new_iemocap_dir, 'test')

os.makedirs(train_dir, exist_ok=True)
os.makedirs(validate_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# Helper function to rename and copy files to new directory structure with the required naming convention
def copy_and_rename_files(file_list, target_dir):
    for file_path in file_list:
        # Get the relative path from the base directory
        rel_path = os.path.relpath(file_path, iemocap_base_dir)

        # Replace slashes with underscores to create the new file name
        new_file_name = rel_path.replace('/', '_')  # Example: "Session1_sentences_wav_Ses01F_script02_1_Ses01F_script02_1_F012.wav"

        # Construct the target path
        target_file_path = os.path.join(target_dir, new_file_name)

        # Ensure target subdirectory exists
        os.makedirs(os.path.dirname(target_file_path), exist_ok=True)

        # Copy the file to the new location with the new name
        shutil.copy2(file_path, target_file_path)

# Copy and rename the files for each set
print(f"Copying and renaming {len(train_files)} training files...")
copy_and_rename_files(train_files, train_dir)

print(f"Copying and renaming {len(validate_files)} validation files...")
copy_and_rename_files(validate_files, validate_dir)

print(f"Copying and renaming {len(test_files)} test files...")
copy_and_rename_files(test_files, test_dir)

print("All files copied and renamed successfully!")
```

```python
import os
import shutil
import random

# Paths to directories
val_dir = '/content/drive/My Drive/MENG/Deep Learning/Project/ASVspoof2019_NewSplit/test'
larger_val_set_dir = '/content/drive/My Drive/MENG/Deep Learning/Project/ASVspoof2019/LA/LA/ASVspoof2019_LA_eval/flac'

# List of files in the val set to be replaced
files_to_replace = [
    'LA_E_A1952273.flac',
    'LA_E_A1895385.flac',
    'LA_E_A2840122.flac',
    'LA_E_A2921021.flac',
    'LA_E_A4301313.flac',
    'LA_E_A4500897.flac',
    'LA_E_A5705432.flac',
    'LA_E_A7643994.flac',
    'LA_E_A7814286.flac',
    'LA_E_A8390905.flac',
    'LA_E_A8882063.flac',
    'LA_E_A8972701.flac',
    'LA_E_A9967055.flac',
]

# Remove files to be replaced from val_dir
for file_name in files_to_replace:
    file_path = os.path.join(val_dir, file_name)
    if os.path.exists(file_path):
        os.remove(file_path)  # Delete the file if it exists
        print(f"Removed {file_name} from val set.")
    else:
        print(f"File {file_name} not found in val set, skipping removal.")

# Get the list of all .flac files in the current smaller val directory (after deletions)
existing_val_files = set(f.replace('.flac', '') for f in os.listdir(val_dir) if f.endswith('.flac'))

# Get all files in the larger val set
all_files_in_larger_val = [f for f in os.listdir(larger_val_set_dir) if f.endswith('.flac')]

# Select replacement files that are not already in the smaller val set
replacement_files = []
while len(replacement_files) < len(files_to_replace):
    candidate = random.choice(all_files_in_larger_val)
    candidate_base = candidate.replace('.flac', '')  # Remove extension for comparison
    if candidate_base not in existing_val_files and candidate not in replacement_files:
        replacement_files.append(candidate)

# Copy the selected replacement files to the val directory
for new_file_name in replacement_files:
    source_path = os.path.join(larger_val_set_dir, new_file_name)
    target_path = os.path.join(val_dir, new_file_name)
    shutil.copy(source_path, target_path)
    print(f"Copied {new_file_name} from larger val set to val set.")

print("\nReplacement process completed.")
```
